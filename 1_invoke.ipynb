```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

# 加载环境变量
load_dotenv()
GROQ_API_KEY = os.getenv("GROQ_API_KEY")

# print(GROQ_API_KEY)
print(load_dotenv())
# 初始化模型
model = init_chat_model("groq:llama-3.3-70b-versatile", api_key=GROQ_API_KEY)
```

```python
# 使用字符串直接调用模型
response = model.invoke("你好！请用一句话介绍什么是人工智能。")

print(response.content)
print(type(response))
print(response)
print("additional_kwargs: ",response.additional_kwargs)
print("response_metadata: ",response.response_metadata)
print("token_usage: ",response.response_metadata["token_usage"])
print("completion_tokens: ",response.response_metadata["token_usage"]["completion_tokens"])
```

```python
# 使用消息列表
messages = [
    SystemMessage(content="你是一个友好的 Python 编程助手，擅长用简单易懂的方式解释编程概念。 回答字数不超过100字。"),
    HumanMessage(content="什么是 Python 装饰器？ "),
]

print("系统提示:", messages[0].content)
print("用户问题:", messages[1].content)
response = model.invoke(messages)
print(response.content)
# 继续对话：将 AI 的回复添加到对话历史
messages.append(response)
print("-------------------------")
print(len(messages))
print(messages[0])
print(messages[1])
print(messages[2])
print("-------------------------")

messages.append(HumanMessage(content="能给我一个简单的例子吗？"))
print(messages)
print(len(messages))

print(messages[3])
print(messages[-1].content)

response2 = model.invoke(messages)
print(response2.content)
```

```python
# 使用字典格式构建消息
messages = [
    {"role": "system", "content": "你是一个专业的数学老师。"},
    {"role": "user", "content": "什么是斐波那契数列？"},
]

print("消息列表:")
for msg in messages:
    print(f"  {msg['role']}: {msg['content']}")

print(messages)
print(messages[0]["role"])
print(messages[0]["content"])
print(messages[1]["role"])
response = model.invoke(messages)

print(response.content)
```

```python
model_deterministic = init_chat_model(
    "groq:llama-3.3-70b-versatile",
    temperature=0,  # 最确定性
    max_tokens=100    # 限制输出长度
)

prompt = "写一个关于春天的句子。"


# 调用两次，观察输出的一致性
for i in range(2):
    response = model_deterministic.invoke(prompt)
    print(f"  第 {i+1} 次: {response.content}")

```

```python
# 创建一个温度较高的模型（更随机）
model_creative = init_chat_model(
    "groq:llama-3.3-70b-versatile",
    temperature=1.5,  # 更有创造性
    max_tokens=100
)


# 调用两次，观察输出的差异
for i in range(2):
    response = model_creative.invoke(prompt)
    print(f"  第 {i+1} 次: {response.content}")
```

```python
response = model.invoke("解释一下什么是递归？用一句话。")


print(response)
print(response.content)


for key, value in response.response_metadata.items():
    print(f"{key}: {value}")

print(f"消息类型: {type(response).__name__}")
print(f"消息 ID: {response.id}")

# 检查 token 使用情况（如果可用）
if "token_usage" in response.response_metadata:
    usage = response.response_metadata["token_usage"]
    print("Token 使用情况:")
    print(f"   提示 tokens: {usage.get('prompt_tokens', 'N/A')}")
    print(f"   完成 tokens: {usage.get('completion_tokens', 'N/A')}")
    print(f"   总计 tokens: {usage.get('total_tokens', 'N/A')}")
```
